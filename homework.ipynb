```python
region_summary = (
    df.groupby("Warehouse_Region")
    .agg(
        total_revenue=("Order_Amount", "sum"),
        avg_units_sold=("Quantity", "mean"),
        num_transactions=("Order_Amount", "count"),
    )
    .reset_index()
    .rename(columns={"Warehouse_Region": "region"})
)

# Region metadata dataframe (including 'Central' present in the dataset)
region_info = pd.DataFrame(
    {
        "region": ["North", "South", "East", "West", "Central"],
        "manager": [
            "Anil Meheta",
            "Priya Rao",
            "Sanjay Gupta",
            "Divya Nair",
            "Amit Shah",
        ],
        "Target_revenue": [2200000, 2200000, 2200000, 2200000, 2200000],
    }
)

# Merge region summary with region info
merged = region_summary.merge(region_info, on="region", how="left")

# Calculate percentage of revenue target achieved
merged["pct_of_target"] = (
    merged["total_revenue"] / merged["Target_revenue"] * 100
)

merged
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_csv("ecommerce_orders_dataset - ecommerce_orders_dataset.csv")

# HISTOGRAM: Distribution of Order Amount
plt.figure(figsize=(8,5))

sns.histplot(df["Order_Amount"], bins=30, kde=True)

plt.title("Distribution of Order Amount", fontsize=14, fontweight="bold")
plt.xlabel("Order Amount ($)")
plt.ylabel("Frequency")

plt.show()
```

```python
#BAR PLOT: Average Order Amount by Warehouse Region
plt.figure(figsize=(10, 6))
sns.barplot(x="Warehouse_Region", y="Order_Amount", data=df)

plt.xticks(rotation=0) # rotation=90 if region names overlap, but 0 is clean here
plt.title("Average Order Amount by Warehouse Region", fontsize=14, fontweight="bold")
plt.xlabel("Warehouse Region", fontsize=12)
plt.ylabel("Average Order Amount ($)", fontsize=12)
plt.show()
```

```python
# SCATTER PLOT: Unit Price vs. Order Amount
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df, x="Unit_Price", y="Order_Amount", alpha=0.6, color="blue"
)

plt.title("Unit Price vs. Order Amount", fontsize=14, fontweight="bold")
plt.xlabel("Unit Price ($)")
plt.ylabel("Order Amount ($)")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# BOX PLOT: Order Amount Distribution across Customer Segments
plt.figure(figsize=(10, 6)) # Create a new figure for the box plot
sns.boxplot(
    data=df, x="Customer_Segment", y="Order_Amount")
plt.title("Order Amount Distribution across Customer Segments", fontsize=14, fontweight="bold")
plt.xlabel("Customer Segment", fontsize=12)
plt.ylabel("Average Order Amount ($)", fontsize=12)
plt.show() # Show the box plot
```
